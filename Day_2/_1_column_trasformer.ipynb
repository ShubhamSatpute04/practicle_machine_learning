```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,minmax_scale,OneHotEncoder,OrdinalEncoder,LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
df=pd.read_csv('covid_toy.csv')
```

```python
si=SimpleImputer()
X=df.drop('has_covid',axis=1)
y=df['has_covid']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=0,test_size=0.2)
```

```python
ct=ColumnTransformer(transformers=[
        ('tf_le',OneHotEncoder(drop='first'),['gender','city']),
        ('tf_oe',OrdinalEncoder(categories=[['Strong','Mild']]),['cough']),
        ('tf_si',SimpleImputer(),['fever'])
    ],remainder='passthrough'
)
```

```python
X_train_transformed=ct.fit_transform(X_train)
X_test_transformed=ct.transform(X_test)
```
