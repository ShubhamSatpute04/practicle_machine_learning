```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline,make_pipeline
from sklearn.feature_selection import SelectKBest,chi2
from sklearn.tree import DecisionTreeClassifier
```

```python
df=pd.read_csv('train.csv')
df.head(2)
```

```python
# name - drop
# sex , embarked- ohe
# cabin , ticket - remove alphabate 
df.dtypes
```

```python
df.info()
```

```python
X=df.drop(['Survived','Name','Ticket','Cabin'],axis=1)
y=df['Survived']
```

```python
X.head()
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=0,test_size=0.2)
```

```python
tr_si1=ColumnTransformer([
    ('imp_age',SimpleImputer(strategy='most_frequent'),[2]),
    ('imp_emb',SimpleImputer(strategy='most_frequent'),[7]),
],remainder='passthrough')
```

```python
tr_ohe2=ColumnTransformer([
    ('ohe_',OneHotEncoder(handle_unknown='ignore'),[2,7])
],remainder='passthrough')
```

```python
tr_scal3=ColumnTransformer([
    ('scl_',MinMaxScaler(),slice(0,7)),
])
```

```python
pipe=make_pipeline(tr_si1,tr_ohe2,tr_scal3)
```

```python
X.sample()
```

```python
pipe.fit(X_train,y_train)
```
