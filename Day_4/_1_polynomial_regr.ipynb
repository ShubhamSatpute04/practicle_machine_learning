```python
# code for generation of polynimial data
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

X = np.sort(4 * np.random.rand(80, 1), axis=0)
y = X.ravel()**2 + np.random.randn(80) * 2

plt.scatter(X, y)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2, random_state=0)
```

```python
# code for linear regression to show that its not performing well on polynomial features
from sklearn.linear_model import LinearRegression
lr= LinearRegression()
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score
```

```python
print(r2_score(y_test,y_pred))
```

```python
# Polynomial regression code starts here 
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2)
x_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
```

```python
lr.fit(x_train_poly,y_train)
```

```python
y_pred=lr.predict(X_test_poly)
```

```python
print(r2_score(y_test,y_pred))
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 100

x1 = np.random.randint(1, 21, n)
x2 = np.random.randint(1, 21, n)
x3 = np.random.randint(1, 21, n)

# Gaussian noise
noise = np.random.normal(0, 20, n)

y = (
    2 * x1**2
    + 3 * x2**2
    + x3**2
    + 2 * x1 * x2
    + 5
    + noise
)

df = pd.DataFrame({
    "x1": x1,
    "x2": x2,
    "x3": x3,
    "y": np.round(y, 2)
})

df.to_csv("multiple_polynomial_regression_noisy.csv", index=False)

print(df.head())
print(f"\nRows: {len(df)}")
```

```python
lr.fit(X_train,y_train)
```

```python
y_pred=lr.predict(X_test)
```
